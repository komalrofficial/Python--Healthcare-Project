```python
df.Blood_Pressure.unique()
```

```python
import numpy as np

conditions = [
    df['Blood_Pressure'] < 90,
    (df['Blood_Pressure'] >= 90) & (df['Blood_Pressure'] <= 120),
    (df['Blood_Pressure'] >= 120) & (df['Blood_Pressure'] <= 179)
]

choices = ['Low', 'Normal', 'High']

df['Blood_Pressure_Category'] = np.select(
    conditions,
    choices,
    default='Unknown'
)

df
```

Line Plot (Average Blood Pressure by Age)

```python
bp_age = df.groupby('Age')['Blood_Pressure'].mean()

bp_age.plot(
    kind='line',
    figsize=(8,4),
    marker='o'
)

plt.title('Average Blood Pressure by Age')
plt.xlabel('Age')
plt.ylabel('Average Blood Pressure')
plt.grid(True)
plt.show()
```

```python
df.Heart_Rate.unique()
```

```python

conditions = [
    df['Heart_Rate'] < 60,
    (df['Heart_Rate'] >= 60) & (df['Heart_Rate'] <= 100),
    df['Heart_Rate'] >= 100
]

choices = ['Low', 'Normal', 'High']

df['Heart_Rate_Category'] = np.select(
    conditions,
    choices,
    default='Unknown')

df
```

Scatter Plot (Age vs Heart Rate)

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,4))

plt.scatter(df['Age'], df['Heart_Rate'])

plt.title('Age vs Heart Rate')
plt.xlabel('Age')
plt.ylabel('Heart Rate')

plt.show()
```

```python
df.Cholesterol_Level.unique()
```

```python
import numpy as np

conditions = [
    df['Cholesterol_Level'] < 200,
    (df['Cholesterol_Level'] >= 200) & (df['Cholesterol_Level'] <= 239),
    df['Cholesterol_Level'] >= 240
]

choices = ['Low', 'Normal', 'High']

df['Cholesterol_Category'] = np.select(
    conditions,
    choices,
    default='Unknown'
)

df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,4))

sns.barplot(
    data=df,
    x='Cholesterol_Category',
    y='Age'
)

plt.title("Average Age by Cholesterol Category")
plt.xlabel("Cholesterol Category")
plt.ylabel("Average Age")

plt.show()
```

```python
pivot_table = pd.pivot_table(
    df,
    index=['Blood_Pressure_Category',
           'Heart_Rate_Category',
           'Cholesterol_Category'],
    columns='Diagnosis',
    aggfunc='size',
    fill_value=0
)

pivot_table
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12,8))

sns.heatmap(
    pivot_table,
    annot=True,
    fmt='d',
    cmap='YlGnBu'
)

plt.title('Diagnosis by BP, Heart Rate and Cholesterol Categories')
plt.xlabel('Diagnosis')
plt.ylabel('BP | Heart Rate | Cholesterol')

plt.show()
```
